```python
def missing_report(df):
    report = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percentage": (
            df.isna().sum() / len(df) * 100
        )
    })

    return report.sort_values(
        "missing_count",
        ascending=False
    )
```

```python
missing_report(c)
```

```python
missing_report(t)
```

```python
expected_revenue = (
    t["quantity"]
    * t["unit_price"]
    * (1 - t["discount_pct"] / 100)
)
```

```python
revenue_difference = (
    t["revenue"]
    - expected_revenue
).abs()
```

```python
revenue_difference.describe()
```
